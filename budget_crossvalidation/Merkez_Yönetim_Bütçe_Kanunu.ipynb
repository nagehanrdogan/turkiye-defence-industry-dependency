```python
import pandas as pd

# Topladığın Merkez Yönetim Bütçe Kanunu başlangıç ödenekleri (geniş format)
data = {
    "Yil": [2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025, 2026],
    "MSB": [40402239000, 46462303000, 53859342000, 61484939000, 80439536000, 182770968000, 440496960000, 623899190000, 822930177000],
    "Jandarma": [13311208000, 19604133000, 22968117000, 27666860000, 35996445000, 76073628000, 189693588000, 261262624000, 375056157000],
    "Sahil_Guvenlik": [682700000, 869313000, 1128824000, 1470792000, 1917657000, 4986900000, 11517337000, 16766308000, 23093023000],
    "SSB": [66011000, 87822000, 100839000, 120063000, 153974000, 344899000, 837052000, 1137681000, 1546524000]
}

df_wide = pd.DataFrame(data)

# Tidy (uzun) formata çevirme
df_tidy = df_wide.melt(id_vars=["Yil"], 
                       value_vars=["MSB", "Jandarma", "Sahil_Guvenlik", "SSB"], 
                       var_name="Kurum", 
                       value_name="Tutar_TL")

# Eksik sütunları ekleme
df_tidy["Odenek_Turu"] = "baslangic_odenegi"
df_tidy["Kaynak"] = df_tidy["Yil"].astype(str) + " MYB Kanunu"

# İlk birkaç satırı kontrol edelim
print(df_tidy.sort_values(by=["Yil", "Kurum"]).head(8))

# İstersen doğrudan CSV olarak kaydedebilirsin:
df_tidy.to_csv("savunma_butce_2018_2026_tidy.csv", index=False)
```
